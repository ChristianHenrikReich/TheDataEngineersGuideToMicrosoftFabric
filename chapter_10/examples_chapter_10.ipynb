Example 10-7: Creating a restore point before loading a table

```python
# Required dependencies
import com.microsoft.spark.fabric
from sempy_labs.warehouse import create_restore_point

# Creating the restore point
create_restore_point(
    warehouse="gold",
    name=f"silver_to_gold",
    description="Restore point before processing Silver to Gold.",
)

# Copying data from silver.dbo.customers to gold
df = spark.table("silver.dbo.customers")
df.write.mode("overwrite").synapsesql("gold.dbo.customers")
```
