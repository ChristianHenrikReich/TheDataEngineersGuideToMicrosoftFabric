Example 6.9: Delta Lake auto compaction Spark session level

```python
spark_settings = [
    "spark.microsoft.delta.optimize.fast.enabled",
    "spark.microsoft.delta.optimize.fileLevelTarget.enabled",
    "spark.databricks.delta.autoCompact.enabled",
]

for setting in spark_settings:
    print(f"{setting} is {spark.conf.get(setting, 'False')}")
```

Example 6.10: Delta Lake auto compaction table level

```python
from pyspark.sql.functions import col, lit

has_auto_compact = (
    spark.sql("SHOW TBLPROPERTIES silver.customers")
    .where(col("key") == lit("delta.autoOptimize.autoCompact"))
    .first()
)

print(f"Has auto compaction: {has_auto_compact.value}")
```
