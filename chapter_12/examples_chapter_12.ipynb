Example 12-7. Incremental bronze-to-silver processing with streaming stopping after
batch

```python
from pyspark.sql import functions as F
bronze_stream_df = spark.readStream.table("orders")
silver_stream_df = bronze_stream_df.where(F.col('total') > 0)

silver_query = (
    silver_stream_df
    .writeStream
    .format("delta")
    .outputMode("append")
    .option("checkpointLocation", "Files/checkpoints/silver_orders")
    .trigger(availableNow=True)
    .toTable("silver.orders")
)

silver_query.awaitTermination()
```
