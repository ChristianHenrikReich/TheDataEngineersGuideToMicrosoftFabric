#### **Due to the way Fabric operates, a top level notebook calling transformations is the best pattern. In this way, all the notebooks can run 1 on when cluster. Running the notebook in other ways will result in, each notebook spins up another cluster. It adds startup penalty as well as cluster exhaustion.**

# **Execute multiple notebooks simultaneously**

####  Load dimensions first and then facts

```python
DAG = { 
    "activities": [
        { 
            "name": "3_Load_Dimension_Sales_Customers", # activity name, must be unique 
            "path": "3_Load_Dimension_Sales_Customers", # notebook path 
            "args": {'useRootDefaultLakehouse': True}, # notebook parameters 
        }, 
        { 
            "name": "3_Load_Dimension_Sales_Products", 
            "path": "3_Load_Dimension_Sales_Products", 
            "args": {'useRootDefaultLakehouse': True} 
        }, 
        { 
            "name": "3_Load_Dimension_Dates", 
            "path": "3_Load_Dimension_Dates", 
            "args": {'useRootDefaultLakehouse': True}, 
        },
        { 
            "name": "4_Load_Fact_Sales_Transactions", 
            "path": "4_Load_Fact_Sales_Transactions", 
            "args": {'useRootDefaultLakehouse': True}, 
            "dependencies": ["3_Load_Dimension_Sales_Customers", "3_Load_Dimension_Sales_Products", "3_Load_Dimension_Dates"] # list of activity names that this activity depends on 
        } 
    ] 
} 

notebookutils.notebook.runMultiple(DAG, {"displayDAGViaGraphviz":True, "DAGLayout":"spectral", "DAGSize":11})
```
