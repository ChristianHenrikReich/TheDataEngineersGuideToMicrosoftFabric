# **Defining business requirements**

#### This could also be pure PySpark, it depends on the skill set of the BI Developer.

```python
%%sql
CREATE OR REPLACE TEMPORARY VIEW dim_sales_customers_df AS

SELECT
    -- Keys
   Id as DWID_Customer,

   -- Business Keys
   CustomerID,

    -- Attributes
   FirstName,
   LastName,
   FirstName + ' ' + LastName AS FullName
FROM
   silver.dbo.sales_customers
```

# **Writing the dimension**

#### Fetch the DataFrame and pass it to our function for writing dimensions

```python
dim_df = spark.table('dim_sales_customers_df')

load_dimension(df=dim_df, lakehouse=destination_lakehouse, destination=dimension_destination)
```
