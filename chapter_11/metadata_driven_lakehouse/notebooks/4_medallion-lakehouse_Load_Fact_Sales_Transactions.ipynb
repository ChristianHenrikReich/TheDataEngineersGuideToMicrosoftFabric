# **Defining business requirements**

#### This could also be pure PySpark, it depends on the skill set of the BI Developer.

```python
%%sql
CREATE OR REPLACE TEMPORARY VIEW fact_sales_transactions_df AS

SELECT
    -- Keys
   CAST(date_format(TransactionDate,'yMMdd') AS INT) AS DWID_Date,
   CAST(CustomerId AS INT) AS DWID_Customer,
   CAST(ProductID AS INT) AS DWID_Product,

    -- Attributes
   TransactionID,

   -- Measures
   CAST(Quantity AS INT) AS M_Quantity,
   TotalPrice / Quantity AS M_UnitPrice,
   TotalPrice / 1 AS M_TotalPrice
FROM
   silver.dbo.sales_transactions
```

# **Writing the fact**

#### Fetch the DataFrame and pass it to our function for writing facts

```python
fact_df = spark.table('fact_sales_transactions_df')

load_fact(df=fact_df, lakehouse=destination_lakehouse, destination=fact_destination)
```
